```python
# Assuming you have a DataFrame named df
def extract_most_specific_category(cat_string):
    if pd.isna(cat_string):
        return "Unknown"
    categories = [c.strip() for c in cat_string.split(',') if c.strip()]
    return categories[-1] if categories else "Unknown"

# Apply function
df['specific_category'] = df['categories'].apply(extract_most_specific_category)
```

```python
sampled_df['text_for_clustering'] = (
    df['brand'].fillna('') + ' ' +
    df['specific_category'].fillna('') + ' ' +
    df['name'].fillna('')
)
category_texts = sampled_df['text_for_clustering'].tolist()
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
category_embeddings = model.encode(category_texts, show_progress_bar=True)
```

```python
from sklearn.cluster import KMeans

k = 5  # adjust as needed
kmeans = KMeans(n_clusters=k, random_state=42)
clusters = kmeans.fit_predict(category_embeddings)
sampled_df['cluster'] = clusters
```

```python
from sklearn.manifold import TSNE
import matplotlib.pyplot as plt
import seaborn as sns

# Reduce embeddings to 2D
tsne = TSNE(n_components=2, random_state=42)
tsne_embeddings = tsne.fit_transform(category_embeddings)

# Add to DataFrame
sampled_df['tsne_x'] = tsne_embeddings[:, 0]
sampled_df['tsne_y'] = tsne_embeddings[:, 1]

# Plot
plt.figure(figsize=(10, 7))
sns.scatterplot(data=sampled_df, x='tsne_x', y='tsne_y', hue='cluster', palette='tab10')
plt.title("t-SNE Visualization of Clusters")
plt.legend(title="Cluster")
plt.show()
```

```python
grouped = sampled_df.groupby('meta_category')
```

```python
def summarize_category(df_cat):
    top_products = (
        df_cat.groupby('name')['reviews.rating']
        .mean()
        .sort_values(ascending=False)
        .head(3)
        .index.tolist()
    )

    worst_product = (
        df_cat.groupby('name')['reviews.rating']
        .mean()
        .sort_values()
        .index[0]
    )

    summaries = {
        'top_products': top_products,
        'worst_product': worst_product,
        'reviews_text': df_cat[df_cat['name'].isin(top_products + [worst_product])]['reviews.text'].tolist()
    }
    return summaries

category_summaries = {cat: summarize_category(group) for cat, group in grouped}
```

```python
from openai import OpenAI
from google.colab import userdata
api_key = userdata.get('OPENAI_API')
client = OpenAI(api_key=api_key)

def get_completion(prompt, model="gpt-4"): # Andrew mentioned that the prompt/ completion paradigm is preferable for this class
    messages = [{"role": "user", "content": prompt}]
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=0, # this is the degree of randomness of the model's output
    )
    return response.choices[0].message.content
```

```python
import random

def build_prompt(category, summary):
    top_products = summary.get('top_products', [])
    worst_product = summary.get('worst_product', "<not available>")

    # Pad top products list if less than 3
    if len(top_products) < 3:
        top_products += ["<not available>"] * (3 - len(top_products))

    # Randomized intro templates
    intros = [
        f"Exploring the best {category} you can buy today, here are the top contenders based on user reviews and performance.",
        f"If you're shopping for {category}, check out these standout choices that are making waves.",
        f"{category} reviews are in, and we've rounded up the top-rated options and one to avoid.",
        f"Let’s dive into the best {category} currently available and what sets them apart."
    ]

    # Randomized key differences phrasing
    key_diff_prompts = [
        "Compare their key features, strengths, and common complaints.",
        "Highlight what makes each product unique and what users tend to dislike.",
        "Discuss the pros and cons of each to help readers decide.",
        "Break down how they differ and what reviewers are saying."
    ]

    # Randomized outro for worst product
    worst_product_prompts = [
        f"Also, take caution with the lowest-rated pick: {worst_product}. Explain what went wrong according to reviewers.",
        f"On the other hand, avoid {worst_product} — it's received poor feedback. Elaborate on the common issues.",
        f"Lastly, beware of {worst_product}, which stands out for all the wrong reasons. Describe why it's not recommended.",
        f"Unfortunately, {worst_product} didn’t make the cut — summarize the main criticisms it received."
    ]

    prompt = f"""
{random.choice(intros)}

🏆 Top 3 Products:
1. {top_products[0]}
2. {top_products[1]}
3. {top_products[2]}

🧐 {random.choice(key_diff_prompts)}

⚠️ {random.choice(worst_product_prompts)}
"""

    return prompt.strip()
```

```python

for category, summary in category_summaries.items():
    prompt = build_prompt(category, summary)
response = get_completion(prompt)
print(response)
```
